```python
import requests
import os
from dotenv import load_dotenv

load_dotenv()

url = 'http://api.opentripmap.com/0.1/en/places/geoname'
api_key = os.environ.get("OPENTRIPMAP_API_KEY")

params = {
    'apikey' : api_key ,
    'lang' : 'en' ,
    'name' : 'Lisbon'
}

response = requests.get(url,params=params)

response=response.json()

response
```

```python
lat = response['lat']
lon = response['lon']
radius = 5000000000
lang = 'en'

url = 'http://api.opentripmap.com/0.1/en/places/radius'

params = {
    'apikey' : api_key ,
    'lang' : lang ,
    'lat' : lat ,
    'lon' : lon ,
    'radius' : radius
}

data = requests.get(url,params=params)

data.json()
```

```python
for results in data.json()['features']:
    
    print(results['properties']['name'])
    print(results['properties']['kinds'].split(','))
    print('------------')

```
