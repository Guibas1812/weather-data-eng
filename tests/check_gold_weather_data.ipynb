```python
from dotenv import load_dotenv
import boto3
from pathlib import Path
import pandas as pd

load_dotenv()  # reads .env, sets env variables

s3 = boto3.client("s3")
bucket_name = "weather-data-eng" #bucket_name

l3y_avg_df = pd.read_csv(s3.get_object(
    Bucket=bucket_name,
    Key="gold/weather/fact_weather_l3y_avg.csv")['Body'])

monthly_by_year_df = pd.read_csv(s3.get_object(
    Bucket=bucket_name,
    Key="gold/weather/fact_weather_monthly_by_year.csv")['Body'])
```
