```python
import requests
from dotenv import load_dotenv
import boto3
from pathlib import Path
import pandas as pd
import json

load_dotenv()  # reads .env, sets env variables

s3 = boto3.client("s3")
bucket_name = "weather-data-eng" #bucket_name

weather_data = pd.read_csv(s3.get_object(
            Bucket=bucket_name,
            Key="silver/weather/daily_data.csv")['Body'])


weather_data.head(20)
```

```python
print("SHAPE (rows, cols):", weather_data.shape)
print("------")

print("INFO:")
print(weather_data.info())
print("------")

print("DESCRIBE:")
print(weather_data.describe())
print("------")

print("NULL COUNTS:")
print(weather_data.isnull().sum())
print("------")

print("DUPLICATE CITY NAMES:", weather_data['city'].duplicated().sum())
print("------")

print("UNIQUE CITIES:", weather_data['city'].nunique())
print(weather_data['city'].unique())
print("------")
```
