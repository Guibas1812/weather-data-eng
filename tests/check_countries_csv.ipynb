```python
from dotenv import load_dotenv
import boto3
from pathlib import Path
import pandas as pd

load_dotenv()  # reads .env, sets env variables

s3 = boto3.client("s3")
bucket_name = "weather-data-eng" #bucket_name

cities_df = pd.read_csv(s3.get_object(
    Bucket=bucket_name,
    Key="bronze/reference/cities.csv")['Body'])
```

```python
cities_df.head(20)
```

```python
print("SHAPE (rows, cols):", cities_df.shape)
print("------")

print("INFO:")
print(cities_df.info())
print("------")

print("DESCRIBE:")
print(cities_df.describe())
print("------")

print("NULL COUNTS:")
print(cities_df.isnull().sum())
print("------")

print("DUPLICATE CITY NAMES:", cities_df['city'].duplicated().sum())
print("------")

print("UNIQUE CITIES:", cities_df['city'].nunique())
print(cities_df['city'].unique())
print("------")
```
